# Sales Prediction Using Machine Learning

## 1. Import Required Libraries

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

## 2. Load and Understand the Dataset

```python
file_path = "advertising.csv"

df = pd.read_csv(file_path)

df = df.loc[:, ~df.columns.str.contains("^Unnamed")]

print("Dataset loaded successfully!\n")

print("First 5 rows:")
print(df.head())

print("\nDataset Shape:", df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nMissing Values:")
print(df.isnull().sum())
```

## 3. Exploratory Data Analysis

```python
plt.figure(figsize=(7, 4))
sns.scatterplot(data=df, x="TV", y="Sales")
plt.title("TV Advertising vs Sales")
plt.show()

plt.figure(figsize=(7, 4))
sns.scatterplot(data=df, x="Radio", y="Sales")
plt.title("Radio Advertising vs Sales")
plt.show()

plt.figure(figsize=(7, 4))
sns.scatterplot(data=df, x="Newspaper", y="Sales")
plt.title("Newspaper Advertising vs Sales")
plt.show()
```

## 4. Feature Selection

```python
X = df[["TV", "Radio", "Newspaper"]]
y = df["Sales"]
```

## 5. Train-Test Split

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("\nTraining data size:", X_train.shape)
print("Testing data size:", X_test.shape)
```

## 6. Model Training

```python
model = LinearRegression()

model.fit(X_train, y_train)

print("\nModel trained successfully!")
```

## 7. Sales Prediction

```python
y_pred = model.predict(X_test)
```

## 8. Model Evaluation

```python
mae = mean_absolute_error(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)

print("\nModel Evaluation:")
print(f"Mean Absolute Error: {mae:.2f}")
print(f"Mean Squared Error: {mse:.2f}")
print(f"R2 Score: {r2:.2f}")
```

## 9. Actual vs Predicted Sales

```python
plt.figure(figsize=(7, 5))

plt.scatter(y_test, y_pred, color="blue")
plt.plot(
    [y_test.min(), y_test.max()],
    [y_test.min(), y_test.max()],
    color="red",
    linewidth=2
)

plt.title("Actual Sales vs Predicted Sales")
plt.xlabel("Actual Sales")
plt.ylabel("Predicted Sales")

plt.show()
```

## 10. Predict Sales for New Advertising Data

```python
new_advertising_data = pd.DataFrame(
    [[150, 25, 30]],
    columns=["TV", "Radio", "Newspaper"]
)

predicted_sales = model.predict(new_advertising_data)

print("\nNew Advertising Budget:")
print(new_advertising_data)

print(f"\nPredicted Sales: {predicted_sales[0]:.2f}")
```

## 11. Conclusion

The Linear Regression model was successfully used to predict product sales based on advertising expenditure across TV, Radio, and Newspaper.

The model achieved an **R² Score of approximately 0.91**, with a **Mean Absolute Error of 1.27** and a **Mean Squared Error of 2.91** on the test dataset.
